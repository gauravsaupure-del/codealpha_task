```python
import matplotlib.pyplot as plt
import seaborn as sns

# Figure size mothi karu jehne karun sagle states clear distil
plt.figure(figsize=(14, 8))

# Data madle extra spaces kadhun taku
df1.columns = df1.columns.str.strip()

# State-wise average unemployment rate calculate karun sort karu
state_unemp = df1.groupby('Region')['Estimated Unemployment Rate (%)'].mean().reset_index()
state_unemp = state_unemp.sort_values(by='Estimated Unemployment Rate (%)', ascending=False)

# Bar plot banvhu (Horizontal bar plot mhanje nave vachtana sopu padel)
sns.barplot(x='Estimated Unemployment Rate (%)', y='Region', data=state_unemp, palette='viridis')

plt.title('Average Unemployment Rate by State in India (COVID-19 Analysis)', fontsize=14, fontweight='bold')
plt.xlabel('Average Estimated Unemployment Rate (%)', fontsize=12)
plt.ylabel('State / Region', fontsize=12)

plt.tight_layout()
plt.show()
```
